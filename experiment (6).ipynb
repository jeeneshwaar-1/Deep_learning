```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import urllib.request
import numpy as np
```

```python
device = torch.device('mps' if torch.backends.mps.is_available() else 'cpu')
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/daily-total-female-births.csv'
data_str = urllib.request.urlopen(url).read().decode('utf-8')
raw_data = [float(line.split(',')[1]) for line in data_str.strip().split('\n')[1:]]
data = (np.array(raw_data) - np.mean(raw_data)) / np.std(raw_data)
```

```python
seq_len = 10
X, y = [], []
for i in range(len(data) - seq_len):
    X.append(data[i:i+seq_len])
    y.append(data[i+seq_len])
X = torch.FloatTensor(X).unsqueeze(-1).to(device)
y = torch.FloatTensor(y).unsqueeze(-1).to(device)
```

```python
class LSTMModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(1, 16, batch_first=True)
        self.fc = nn.Linear(16, 1)
    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])
```

```python
model = LSTMModel().to(device)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
for epoch in range(50):
    optimizer.zero_grad()
    out = model(X)
    loss = criterion(out, y)
    loss.backward()
    optimizer.step()
```

```python
preds = model(X).cpu().detach().numpy()
plt.plot(y.cpu().numpy(), label='True')
plt.plot(preds, label='Pred')
plt.legend()
plt.title('LSTM Predictions on Female Births')
plt.savefig('lstm_predictions.png')
```
